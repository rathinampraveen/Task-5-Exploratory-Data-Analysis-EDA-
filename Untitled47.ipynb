**Task 5: Exploratory Data Analysis (EDA)**

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


```

```python
# Load the dataset
train_df = pd.read_csv("/content/gender_submission.csv")
train_df = pd.read_csv("/content/test.csv")
train_df = pd.read_csv("/content/train.csv")
```

```python
# Display basic info and statistics
print(train_df.info())
print(train_df.describe(include='all'))
print(train_df['Sex'].value_counts())
print(train_df['Embarked'].value_counts())
```

```python
# Histograms of numeric columns
numeric_cols = ['Age', 'Fare', 'SibSp', 'Parch']
train_df[numeric_cols].hist(bins=20, figsize=(14, 8), color='skyblue', edgecolor='black')
plt.suptitle("Histograms of Numeric Features", fontsize=16)
plt.tight_layout()
plt.show()
```

```python
# Boxplots: Age and Fare vs Survived
plt.figure(figsize=(14, 6))
```

```python
plt.subplot(1, 2, 1)
sns.boxplot(x='Survived', y='Age', data=train_df, palette='Set2')
plt.title('Age vs Survived')
```

```python
plt.subplot(1, 2, 2)
sns.boxplot(x='Survived', y='Fare', data=train_df, palette='Set2')
plt.title('Fare vs Survived')
```

```python
plt.tight_layout()
plt.show()
```

```python
# Correlation Heatmap
plt.figure(figsize=(10, 6))
corr = train_df.corr(numeric_only=True)
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()
```

```python
# Pairplot for selected features
selected_features = ['Survived', 'Pclass', 'Age', 'Fare', 'SibSp', 'Parch']
sns.pairplot(train_df[selected_features], hue='Survived', palette='husl')
plt.suptitle("Pairplot of Selected Features", y=1.02)
plt.show()
```
